# Aufgabe 2.1.1
1-D Furier - Basisfunktion , Orthogonalität

Die 1-D Furier - Basisfunktion ist geggeben durch:

$$
\varphi_n(x) = e(iq_nx)
$$
wobei $ x \in [0, L]$ und $q_n = 2\pi n / L$.


Ist das Skalarproduktgleich von zwei Basisvektoren gleich Null, stehen diese Orthogonal zueinander.
Wir benutzen in unserem Fall $\varphi_n , \varphi_m$.

$$
(\varphi_n,\varphi_m) = \frac{1}{L} \int_0^L \varphi_n^*  \varphi_m dx
$$

$$(\varphi_n , \varphi_m )  = \frac{1}{L} \int_0^L e^{-i\frac{2 \pi n}{L}x} \cdot e^{i\frac{2 \pi m}{L}x} dx$$

$$(\varphi_n , \varphi_m ) = \frac{1}{L} \int_0^L e^{i\cdot(m-n)\cdot\frac{2 \pi x}{L}} dx$$


Sollte $m = n$ erhalten wird als Lösung die 1.

Sollte $m \neq n$ erhalten wir als Lösung die 0. Dies ist nachfolgend aufgezeigt:


$$(\varphi_n , \varphi_m )  = \frac{1}{L} \int_0^L e^{i\frac{2 \pi k}{L}x} dx \\
$$

$$
(\varphi_n , \varphi_m ) = \frac{1}{L} ( \int_0^L cos(\frac{2 \pi k}{L}x ) dx + \int_0^L i sin(\frac{2 \pi k}{L}x) dx) \\
$$


Das Integral des cos mit den grenzen 0 , L, wird zu 0 und das Integral des sin mit den Grenzen wird 0,L, jeweils zu 1, daher erhalten wir:

 $$(\varphi_n , \varphi_m ) =  \frac{1}{L} (0 -  0 +  1i -1i) =  0 $$

2D-Basisfunktion:
$$
\varphi_{nm}(x,y) = e^{iq_nx + i q_y} \\
$$
$$
\varphi_{nm}(x,y) = e^{iq_nx} \cdot e^{iq_my}y) \\
\varphi_{nm}(x,y9 = \varphi_n(x) \cdot \varphi_m(y)\\
$$
$q_n$ = $2\pi n / L_x$ $\space\space\space\space$ $q_m$ = $2\pi m / L_y$ $\space\space\space$ $n \in \mathbb{N}$  $\space$ &  $\space$ $m \in \mathbb{N}$
$\\$

Das Sklarprodukt aus der Basisfunktion $\varphi_{jk}(x,y)$ und $\varphi_{nm}(x,y)$ ist:

$$
(\varphi_{nm}(x,y),\varphi_{jk}(x,y) ) = \frac{1}{L_xL_y} \int_0^{L_y} \int_0^{L_x} \varphi_{nm}^* \cdot \varphi_{jk} \space dxdy
$$
$$
(\varphi_{nm}(x,y),\varphi_{jk}(x,y) )= \frac{1}{L_xL_y} \int_0^{L_y} \int_0^{L_x}   \varphi_n(x)^* \cdot \varphi_m(y)^* \cdot   \varphi_j(x) \cdot \varphi_k(y) \space dxdy \\
$$

$$
(\varphi_{nm}(x,y),\varphi_{jk}(x,y) )= \frac{1}{L_xL_y} \int_0^{L_y} \int_0^{L_x}   \varphi_n(x)^* \cdot  \varphi_j(x)  \cdot   \varphi_m(y)^* \cdot \varphi_k(y) \space dxdy \\
$$

Nun $ \varphi_n(x)^* \cdot  \varphi_j(x)$ zu  $\delta_{nj}$ zusammengefasst und auch $\varphi_m(y)^* \cdot \varphi_k(y)$ zu $\delta_{mk}$.

$$
(\varphi_{nm}(x,y),\varphi_{jk}(x,y) ) = \frac{1}{L_xL_y} \int_0^{L_y} \int_0^{L_x} \delta_{nj} \cdot \delta_{mk} ~dxdy = \delta_{nj}  \delta_{mk}
$$
